3. Implement a method `print_list()` that traverses and prints the list.

```python
class LinkedList():
    def __init__(self):
        self.head = None
        
    def print_list(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")
```

4. Implement a method `length()` that returns the number of nodes.

```python
class LinkedList():
    def __init__(self):
        self.head = None
        
    def length(self):
        current = self.head
        len = 0
        while current:
            current = current.next
            len += 1
        return len
    
node0 = Node(10)
node1 = Node(11)
node2 = Node(12)

node0.next = node1
node1.next = node2

linkedlist = LinkedList()
linkedlist.head = node0

print(linkedlist.length())
```

5. Implement `search(value)` that returns `True` if the value exists in the list.

```python
class LinkedList():
    def __init__(self):
        self.head: Node = None
        
    def search(self, value):
        current = self.head
        
        if not current:
            return False
        
        while current:
            if current.data == value:
                return True
            current = current.next
        return False
    
node0 = Node(10)
node1 = Node(11)
node2 = Node(12)

node0.next = node1
node1.next = node2

linkedlist = LinkedList()
linkedlist.head = node0

print(linkedlist.search(10))
```

8. Implement `insert_at_index(index, data)` (0-based index).

```python
class LinkedList():
    def __init__(self):
        self.head: Node = None
        
    def print_list(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")
        
    def insert_at_beginning(self, data):
        old_head = self.head
        head = Node(data)
        self.head = head
        head.next = old_head
        
    def length(self):
        current = self.head
        len = 0
        while current:
            current = current.next
            len += 1
        return len
        
    def insert_at_index(self, index, data):
        
        len = self.length()
        
        if index < 0:
            raise IndexError("Index must be positive")
        
        elif index == 0:
            self.insert_at_beginning(data=data)
            
        elif index > len:
            raise IndexError("Index out of bounds")
        else:
            count = 0
            current = self.head
            
            while current and count < index - 1:
                current = current.next
                count += 1
                
            node = Node(data)
            next_node = current.next
            current.next = node
            node.next = next_node
        
node0 = Node(10)
node1 = Node(11)
node2 = Node(12)

node0.next = node1
node1.next = node2

linkedlist = LinkedList()
linkedlist.head = node0

linkedlist.print_list()
linkedlist.insert_at_index(3, 13)
linkedlist.print_list()
```
